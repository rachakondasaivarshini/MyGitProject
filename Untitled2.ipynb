```python
#GAN-1,2,4,5-EXPS
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Flatten, Dense

(x,y),(xt,yt) = tf.keras.datasets.mnist.load_data()
x,xt = x/255.0,xt/255.0

model = Sequential([
    Flatten(input_shape=(28,28)),
    Dense(128,activation='relu'),
    Dense(10,activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

model.fit(x,y,epochs=1)
model.evaluate(xt,yt)
print("Completed")
```

```python
#GAN-3-EXP
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = np.zeros((200, 200), dtype=np.uint8)
cv2.rectangle(img, (50, 50), (150, 150), 255, -1)

edge = cv2.Canny(img, 100, 200)

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")

plt.subplot(1, 2, 2)
plt.imshow(edge, cmap='gray')
plt.title("Edge Detection")

plt.show()
```

```python
#DS-2,6,7 exps
import pandas as pd
import matplotlib.pyplot as plt

# Create dataset
df = pd.DataFrame({
    'Name': ['A', 'B', 'C', 'D'],
    'Marks': [70, 80, 90, 60],
    'Hours': [2, 3, 4, 1]
})

# Display data
print("Dataset:")
print(df)

# Plot bar chart
df.plot(x='Name', y='Marks', kind='bar')
plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()

# Correlation analysis
print("Correlation Matrix:")
print(df[['Marks', 'Hours']].corr())
```

```python
#DS-3,5 exps
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import MinMaxScaler

# Create dataset
df = pd.DataFrame({
    'Age': [20, 30, None, 40],
    'Marks': [60, 80, 70, 100]
})

print("Original Data:")
print(df)

# Handle missing values
imputer = SimpleImputer(strategy='mean')
data = imputer.fit_transform(df)

# Normalize data
scaler = MinMaxScaler()
result = scaler.fit_transform(data)

print("Transformed Data:")
print(result)
```

```python
#DS-4 exp
import numpy as np
from scipy.spatial.distance import euclidean, cosine, cityblock

# Define vectors
A = np.array([1, 2, 3])
B = np.array([2, 3, 4])

# Calculate distances
print("Euclidean Distance:", euclidean(A, B))
print("Manhattan Distance:", cityblock(A, B))

# Calculate cosine similarity
similarity = 1 - cosine(A, B)
print("Cosine Similarity:", similarity)
```

```python
#DS-8exp
import numpy as np
from scipy.linalg import lu, cholesky, svd

# Define matrix
A = np.array([[4, 2], [2, 3]])

# Matrix operations
print("Addition:\n", A + A)
print("Multiplication:\n", A @ A)

# Eigenvalues
values, vectors = np.linalg.eig(A)
print("Eigenvalues:", values)

# LU decomposition
P, L, U = lu(A)
print("L Matrix:\n", L)
print("U Matrix:\n", U)

# Cholesky decomposition
print("Cholesky:\n", cholesky(A))

# Singular Value Decomposition
u, s, vt = svd(A)
print("SVD Values:", s)
```
