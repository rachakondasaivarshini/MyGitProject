```python
#GAN-common code(1,8)
import tensorflow as tf
from tensorflow.keras import layers, Sequential
import numpy as np
import matplotlib.pyplot as plt

# Load dataset
(x, y), _ = tf.keras.datasets.mnist.load_data()
x = x / 255.0

# Create model
model = Sequential([
    layers.Flatten(input_shape=(28, 28)),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])

# Compile model
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Train model
model.fit(x, y, epochs=1)

# Predict
print("Prediction:", model.predict(x[:1]).argmax())

# Display image
plt.imshow(x[0], cmap='gray')
plt.show()
```

```python
#ds-common code(2,8)
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Create dataset
data = pd.DataFrame({
    'A': [10, 20, 30, 40],
    'B': [20, 30, 40, 50]
})

# Display data
print(data)

# Data preprocessing
scaled = StandardScaler().fit_transform(data)
print("Scaled Data:", scaled)

# Correlation
print("Correlation:", data.corr())

# Visualization
data.plot(kind='bar')
plt.show()

# Distance
print("Distance:", np.linalg.norm(data.iloc[0] - data.iloc[1]))

# PCA
pca = PCA(n_components=1)
print("PCA:", pca.fit_transform(scaled))

# Matrix operations
A = np.array([[4, 2], [2, 3]])
print("Eigenvalues:", np.linalg.eigvals(A))
```
